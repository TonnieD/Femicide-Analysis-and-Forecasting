# Femicide in Kenya: Temporal and Spatial Analysis
![image](../images/crime-scene.JPG)
## Project Objective
In a 2024 NTV article published on November 12, 2024  
[(article link)](https://ntvkenya.co.ke/news/a-woman-is-killed-every-24-hours-kenyas-femicide-nightmare/),  
Kenya’s Deputy Inspector General (DIG) of Police Eliud Lagat reported that **97 femicide cases were recorded in just three months**—approximately **one woman killed every 24 hours**.

This project uses reported femicide data to examine whether such patterns are reflected
in historical trends and how femicide cases evolve **over time and across space** in Kenya.

Specifically, the project seeks to:

- Analyze national femicide trends over time  
- Identify seasonal patterns in reported cases  
- Examine geographic variation across counties  

The analysis begins at the **national level** and progressively zooms into
**county-level temporal patterns**, forming the foundation for future forecasting
and spatial simulations.

---

## Why This Matters

Femicide is not only a criminal issue but a profound **public health, human rights,
and governance concern**.

Understanding **when**, **where**, and **how frequently** femicide occurs enables:

- Evidence-based policy formulation  
- Targeted prevention and intervention strategies  
- Stronger advocacy grounded in data  
- Improved public awareness and accountability  

Time-based and spatial analyses help move the conversation from isolated incidents
to **systemic patterns**.


## Data Description

The dataset consists of reported femicide cases in Kenya, compiled from media articles.
Each row represents a **single reported incident**.

### Key Variables Used in This Stage

- `date of murder`: Date when the incident occurred  
- `county`: County where the incident took place  

Additional variables such as victim age, relationship to suspect, type of killing,
and justice outcomes are retained for later **descriptive and justice-focused analyses**,
but are **not included in the time series modeling stage**.

---

## Data Preparation and Assumptions

- Records with valid murder dates are retained for time series analysis  
- Missing or unknown values in non-temporal variables are preserved as `"Unknown"`  
- Individual cases are aggregated into **monthly counts** to create a consistent time series  
- Monthly aggregation balances noise reduction with the ability to detect meaningful trends  

This results in a national-level monthly time series suitable for trend, seasonality,
and forecasting analysis.


## Methodological Approach

This project adopts a **step-by-step analytical approach**:

1. **National-level time series analysis**  
   - Trend identification  
   - Seasonality assessment  
   - Decomposition of temporal components  

2. **County-level temporal analysis**  
   - Comparison of trends across counties  
   - Identification of regional hotspots  
   - Preparation for spatial simulations  

3. **Extended analysis (later stages)**  
   - Time series forecasting  
   - Interactive dashboards  
   - Contextual and justice-outcome analysis
