# Femicide in Kenya: Temporal and Spatial Analysis
![image](../images/crime-scene.JPG)
## Project Objective
In a 2024 NTV article psoted on November 12, 2024 - [(article link)](https://ntvkenya.co.ke/news/a-woman-is-killed-every-24-hours-kenyas-femicide-nightmare/) it says that "According to Deputy Inspector General (DIG) of Police Eliud Lagat, law enforcement has recorded 97 femicide cases in just the past three months. That’s roughly one woman killed every day."

This project analyzes femicide cases in Kenya to understand:
- National trends over time
- Seasonal patterns in reported cases
- Geographic distribution across counties

The analysis begins at a national level and progressively zooms into county-level patterns.
Later stages of the project will include:
- Time series forecasting
- Interactive dashboards for visualization and exploration

## Why This Matters
Femicide is a critical social issue. Understanding when and where cases occur
can support evidence-based policy, advocacy, and prevention strategies.


## Data Description

The dataset contains reported femicide cases in Kenya, sourced from media articles.
Each row represents a single reported case.

Key variables used in this analysis include:
- `date of murder`: Date when the incident occurred
- `county`: County where the incident took place

Other variables (e.g. age, relationship, verdict) are used in descriptive analysis
but are not included in the time series modeling stage.

For time series analysis, cases are aggregated to monthly counts.
